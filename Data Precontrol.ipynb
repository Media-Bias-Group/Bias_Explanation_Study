```python
import pandas as pd

# Load the CSV file
df = pd.read_csv('results-survey-final-cleaned.csv', delimiter=";")

# Select the 14th column (index 13, as indexing starts from 0)
column_14 = df.iloc[:, 13]

# Count how often each number appears
counts = column_14.value_counts().sort_index()

# Print the result
print("Frequency of each number in column 14:")
print(counts)
```
